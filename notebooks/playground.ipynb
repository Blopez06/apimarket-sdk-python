# API Market SDK Python

Playground para explorar el SDK de forma interactiva.

**Requisito:** Tener una API key de [API Market](https://apimarket.mx).

## Configuración

Antes de hacer cualquier llamada es necesario inicializar el cliente con tu API key.
Puedes pasarla directamente o definir la variable de entorno `APIMARKET_API_KEY`.

```python
import apimarket

# Opción 1: pasar la key directamente
apimarket.assemble(api_key="YOUR_API_KEY_HERE")

# Opción 2: leer de variable de entorno APIMARKET_API_KEY (ya configurada)
# apimarket.assemble()
```

## Cliente síncrono

Por defecto el cliente es síncrono — ideal para scripts y uso general.

```python
import apimarket

apimarket.assemble(api_key="YOUR_API_KEY_HERE")

response = apimarket.fetch_curp_details("LOOA531113HTCPBN07")
print(response.success)
print(response.data.nombres)
print(response.data.apellido_paterno)
```

## Cliente asíncrono

Usa `async_client=True` para obtener corrutinas — útil dentro de aplicaciones async o frameworks como FastAPI.

```python
import apimarket

apimarket.assemble(api_key="YOUR_API_KEY_HERE", async_client=True)

response = await apimarket.fetch_curp_details("LOOA531113HTCPBN07")
print(response.success)
print(response.data.nombres)
```

## Sandbox

Activa el modo sandbox para pruebas — las respuestas son simuladas y no consumen créditos.

```python
import apimarket

apimarket.assemble(api_key="YOUR_API_KEY_HERE", sandbox=True)

response = apimarket.fetch_curp_details("LOOA531113HTCPBN07")
print(response.message)  # 'Esta es una peticion de prueba'
```

## Manejo de errores de validación

El SDK valida los datos antes de llamar a la API y lanza excepciones con códigos descriptivos.

```python
import apimarket
from apimarket.validations import InvalidCURPError

apimarket.assemble(api_key="YOUR_API_KEY_HERE")

try:
    apimarket.fetch_curp_details("INVALIDA")
except InvalidCURPError as e:
    print(e.code.value)   # INVALID_CURP_LENGTH
    print(e.message)      # Longitud inválida: se esperaban 18 caracteres, se recibieron 7.
```
