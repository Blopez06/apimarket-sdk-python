# API Market SDK Python

Playground para explorar el SDK de forma interactiva.

**Requisito:** Tener una API key de [API Market](https://apimarket.mx).

## Configuración

Ejecuta esta celda primero. Lee la API key de la variable de entorno `APIMARKET_API_KEY`.

```python
import apimarket
from apimarket.validations import (
    InvalidCURPError, InvalidNSSError, InvalidRFCError,
    InvalidFolioError, InvalidBirthDateError
)

# Errores de validación agrupados para reutilizar en cada celda
_VALIDATION_ERRORS = (InvalidCURPError, InvalidNSSError, InvalidRFCError, InvalidFolioError, InvalidBirthDateError)

# Opción 1: leer de variable de entorno APIMARKET_API_KEY (recomendado)
apimarket.assemble()

# Opción 2: pasar la key directamente
# apimarket.assemble(api_key="TU_API_KEY")
```

## Cliente síncrono

Por defecto el cliente es síncrono — ideal para scripts y uso general.

```python
try:
    response = apimarket.fetch_curp_details("TEST000101HDFXXX03")
    if response.success and response.data:
        print(response.data.nombres)
        print(response.data.apellido_paterno)
    else:
        print(f"Sin resultados: {response.message}")
except _VALIDATION_ERRORS as e:
    print(f"Error de validación [{e.code.value}]: {e.message}")
```

## Cliente asíncrono

Usa `async_client=True` para obtener corrutinas — útil dentro de aplicaciones async o frameworks como FastAPI.

```python
apimarket.assemble(async_client=True)

try:
    response = await apimarket.fetch_curp_details("TEST000101HDFXXX03")
    if response.success and response.data:
        print(response.data.nombres)
        print(response.data.apellido_paterno)
        print(response.data.fecha_nacimiento)
        print(response.data.sexo)
    else:
        print(f"Sin resultados: {response.message}")
except _VALIDATION_ERRORS as e:
    print(f"Error de validación [{e.code.value}]: {e.message}")
```

## Sandbox

Activa el modo sandbox para pruebas — las respuestas son simuladas y no consumen créditos.

```python
apimarket.assemble(sandbox=True)

try:
    response = apimarket.fetch_curp_details("TEST000101HDFXXX03")
    print(response.message)  # 'Esta es una peticion de prueba'
except _VALIDATION_ERRORS as e:
    print(f"Error de validación [{e.code.value}]: {e.message}")
```

## Manejo de errores de validación

El SDK valida los datos antes de llamar a la API y lanza excepciones con códigos descriptivos.

```python
try:
    apimarket.fetch_curp_details("INVALIDA")
except InvalidCURPError as e:
    print(e.code.value)   # INVALID_CURP_LENGTH
    print(e.message)      # Longitud inválida: se esperaban 18 caracteres, se recibieron 8.
```
